```python
import statsmodels.api as sm
import pandas as pd
import matplotlib.pyplot as plt

from skforecast.plot import set_dark_theme
set_dark_theme()
```

```python
df_serie = sm.datasets.get_rdataset('airquality').data
df_serie['year'] = 1973
df_serie['date'] = pd.to_datetime(df_serie[['year', 'Month', 'Day']])
df_serie.index = df_serie['date']
df_serie.drop(columns=['year', 'Month', 'Day', 'date'], inplace=True)
df_serie
```

```python
plt.figure(figsize=(10, 5))
df_serie['Solar.R'].plot(marker='o')
plt.show()
```

## Imputacion por llenado hacia adelante

```python
plt.figure(figsize=(10, 5))
df_serie['Solar.R'].plot(marker='o')
df_serie['Solar.R'].ffill().plot(alpha=0.4, color='y', ls='dashed', marker='x')
plt.show()
```

## Imputacion por llenado hacia atras

```python
plt.figure(figsize=(10, 5))
df_serie['Solar.R'].plot(marker='o')
df_serie['Solar.R'].bfill().plot(alpha=0.4, color='y', ls='dashed', marker='x')
plt.show()
```

## Imputacion por interpolacion

### Imputacion por interpolacion por cercania

```python
plt.figure(figsize=(10, 5))
df_serie['Solar.R'].plot(marker='o')
df_serie['Solar.R'].interpolate(method='nearest').plot(alpha=0.4, color='y', ls='dashed', marker='x')
plt.show()
```

### Imputacion por interpolacion lineal

```python
plt.figure(figsize=(10, 5))
df_serie['Solar.R'].plot(marker='o')
df_serie['Solar.R'].interpolate(method='linear').plot(alpha=0.4, color='y', ls='dashed', marker='x')
plt.show()
```

### Imputacion por interpolacion de orden superior suavizada

```python
plt.figure(figsize=(10, 5))
df_serie['Solar.R'].plot(marker='o')
df_serie['Solar.R'].interpolate(method='akima').plot(alpha=0.4, color='y', ls='dashed', marker='x')
plt.show()
```
